## Back to the main plot!

We have a LinkedIn profile in variable `linkedin`

We have a summary in variable `summary`

Let's construct a System Prompt..

````python
system_prompt = f"""

# Your role

You are a digital twin running on a website, chatting with visitors of the website.
You represent the person who's website you are on.
You answer questions related to their career, background, skills and experience.

Here are the details of the person you are representing:

{summary}

If asked, you explain clearly that you are an AI that is the digital twin of this person.

# Context

Here is a summary of the person's LinkedIn profile so that you can answer questions:

{linkedin}

# Rules

Engage with the user. Be professional and engaging, as if talking to a potential client or future employer who came across the website.
Avoid answering questions that are not related to the user's career, background, skills and experience;
steer the conversation back to professional topics.

Always stay in character as the digital twin of the person you are representing. Represent the person.

IMPORTANT: If you don't know the answer, say so. Never make up an answer.
If the user asks about something not in the context, say that you don't know.
Never use Markdown formatting such as bold text (**), italics (* or _), or headings (#) in your responses. Always return plain text.
"""
````

````python
display(Markdown(system_prompt))
````

````python
messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": "Hi - please tell me about yourself"},
]
````

````python
response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages)
display(Markdown(response.choices[0].message.content))
````

````python
# Definiëer een function die een chat-gesprek uitvoert
def chat(message, history):
    # Maak een lijst met de boodschappen van het gesprek, beginnend met de system-prompt
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    # return "You said: " + message (Grapje)
    # Maak een aanvraag naar OpenAI's complete gesprekssysteem
    response = openai.chat.completions.create(model="gpt-5.4-nano", messages=messages)

    # Haal de eerste keuze uit de response en lees de inhoud ervan
    return response.choices[0].message.content
````

# And now - TOOLS!

Let's start with a function...

````python
# Definiëer een function die een e-mail record maakt
def record_email_tool(email):
    # Laat weten dat de tool wordt aangeroepen om een e-mail te recorden
    print(f"Tool called to record an e-mail: {email}")

    # Maak een nieuw bestand op in de map "emails.txt" en open het voor het schrijven
    with open("emails.txt", "a", encoding="utf-8") as f:
        # Voeg de aangestuurde e-mail toe aan het bestand, gefolied door een nieuwe regel
        f.write(email + "\n")

    # Returneer een succesbericht
    return "Email ontvangen"
````

## Step 1 - write some json to describe the tool


````python
record_email_tool_json = {
    "name": "record_email_tool",
    "description": "Use this tool to record that a user provided their email address",
    "parameters": {
        "type": "object",
        "properties": {
            "email": {"type": "string", "description": "The email address of this user"}
        },
        "required": ["email"],
        "additionalProperties": False
    }
}
````

````python
tools = [{"type": "function", "function": record_email_tool_json}]
````

````python
tools
````

## Step 2 - a new chat() function

This is where we implement the tool call.

The reality is, it's a bit clunky. This is like seeing the ingredients of a fine recipe, and finding that the ingredients turn out to be quite ordinary.

Tool calling is an "if" statement. In this case, we're hardcoding everything to assume that the only tool is an email tool.

SIDENOTE: If you're thinking - but wait! I should be remembering this so I can do it myself! Then the key point is: this is what Agent Frameworks take care of for you. In practice, you'll likely never type this again yourself. We are shielded from these if statements by the Agent Framework. That's why they're often described as "abstraction layers".

````python
def chat(message, history):
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages, tools=tools)
         
    if response.choices[0].finish_reason=="tool_calls":
            message = response.choices[0].message
            tool_call = message.tool_calls[0]
            email = json.loads(tool_call.function.arguments).get("email")
            record_email_tool(email)
            messages.append(message)
            messages.append({"role": "tool", "content": "Email recorded", "tool_call_id": tool_call.id})
            response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages, tools=tools)
            
    return response.choices[0].message.content
````

````python
gr.ChatInterface(chat).launch(inbrowser=True)
````

## Step 3

Our first ever Agent Loop, done without an Agent Framework!

Changes:
1. Instead of always assuming there's only 1 tool call, iterate through the tools with a for loop
2. Changed from `if finish_reason=="tool_calls"` to `while finish_reason=="tool_calls"`

````python
def chat(message, history):
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages, tools=tools)
         
    while response.choices[0].finish_reason=="tool_calls":
            message = response.choices[0].message
            messages.append(message)
            for tool_call in message.tool_calls:
                email = json.loads(tool_call.function.arguments).get("email")
                record_email_tool(email)
                messages.append({"role": "tool", "content": "Email recorded", "tool_call_id": tool_call.id})
            response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages, tools=tools)
            
    return response.choices[0].message.content
````

````python
gr.ChatInterface(chat).launch(inbrowser=True)
````

````python
record_email_tool_json = {
    "name": "record_email_tool",
    "description": "Use this tool to record that a user provided their email address",
    "parameters": {
        "type": "object",
        "properties": {
            "email": {
                "type": "string",
                "description": "The email address of this user",
            }
        },
        "required": ["email"],
        "additionalProperties": False,
    },
}

# --- NIEUW: Tool definitie voor de evaluator LLM ---
evaluator_tool_json = {
    "name": "evaluate_response_tool",
    "description": "Evaluates if the primary chatbot reply is factual, accurate based on context, and relevant to work.",
    "parameters": {
        "type": "object",
        "properties": {
            "is_valid": {
                "type": "boolean",
                "description": "True if the response is correct and work-related, False otherwise.",
            },
            "reason": {
                "type": "string",
                "description": "Explanation of why the reply is valid or invalid.",
            },
            "corrected_response": {
                "type": "string",
                "description": "Corrected response if is_valid is False. Empty string if True.",
            },
        },
        "required": ["is_valid", "reason"],
        "additionalProperties": False,
    },
}

# Tools voor de primaire chatbot
tools = [{"type": "function", "function": record_email_tool_json}]

# Tool voor de evaluatie-call
evaluator_tools = [{"type": "function", "function": evaluator_tool_json}]
````

````python
import json


def evaluate_reply(user_message, draft_reply, source_context=""):
    eval_messages = [
        {
            "role": "system",
            "content": "Je bent een streng controle-algoritme. Beoordeel of het voorgestelde antwoord klopt met de feiten en werk-gerelateerd is.",
        },
        {
            "role": "user",
            "content": f"Context/Feiten: {source_context}\n\nVraag: {user_message}\n\nVoorgesteld Antwoord: {draft_reply}",
        },
    ]

    eval_response = openai.chat.completions.create(
        model="gpt-5.4-mini",
        messages=eval_messages,
        tools=evaluator_tools,
        tool_choice={
            "type": "function",
            "function": {"name": "evaluate_response_tool"},
        },  # Dwingt het gebruik van de tool af
    )

    # Haal de argumenten uit de tool call
    tool_call = eval_response.choices[0].message.tool_calls[0]
    result = json.loads(tool_call.function.arguments)

    return result
````

````python
def chat(message, history):
    messages = [{"role": "system", "content": system_prompt}] + history + [
        {"role": "user", "content": message}
    ]

    response = openai.chat.completions.create(
        model="gpt-5.4-mini", messages=messages, tools=tools
    )

    while response.choices[0].finish_reason == "tool_calls":
        msg = response.choices[0].message
        messages.append(msg)
        for tool_call in msg.tool_calls:
            email = json.loads(tool_call.function.arguments).get("email")
            record_email_tool(email)
            messages.append(
                {
                    "role": "tool",
                    "content": "Email recorded",
                    "tool_call_id": tool_call.id,
                }
            )
        response = openai.chat.completions.create(
            model="gpt-5.4-mini", messages=messages, tools=tools
        )

    draft_reply = response.choices[0].message.content

    # Controle stap uitvoeren
    eval_result = evaluate_reply(message, draft_reply)

    if eval_result.get("is_valid"):
        return draft_reply
    else:
        # Als het antwoord niet klopt of niet over werk gaat:
        return eval_result.get(
            "corrected_response",
            "Excuses, ik kan alleen vragen beantwoorden die over werk gaan en overeenkomen met onze data.",
        )
````

````python
def chat(message, history):
    messages = [{"role": "system", "content": system_prompt}] + history + [
        {"role": "user", "content": message}
    ]
    response = openai.chat.completions.create(
        model="gpt-5.4-mini", messages=messages, tools=tools
    )

    while response.choices[0].finish_reason == "tool_calls":
        # AANPASSING 1: Gebruik 'response_msg' i.p.v. 'message' om de originele 'message' niet te overschrijven
        response_msg = response.choices[0].message
        messages.append(response_msg)

        for tool_call in response_msg.tool_calls:
            email = json.loads(tool_call.function.arguments).get("email")
            record_email_tool(email)
            messages.append(
                {
                    "role": "tool",
                    "content": "Email recorded",
                    "tool_call_id": tool_call.id,
                }
            )

        response = openai.chat.completions.create(
            model="gpt-5.4-mini", messages=messages, tools=tools
        )

    # Dit is het concept-antwoord van je primaire chatbot
    draft_reply = response.choices[0].message.content

    # AANPASSING 2: Voeg hier de controle toe!
    eval_result = evaluate_reply(user_message=message, draft_reply=draft_reply)

    if eval_result.get("is_valid"):
        return draft_reply
    else:
        return eval_result.get(
            "corrected_response",
            "Ik kan alleen vragen beantwoorden die werkgerelateerd zijn en overeenkomen met onze gegevens.",
        )
````

````python
gr.ChatInterface(chat).launch(inbrowser=True)
````
